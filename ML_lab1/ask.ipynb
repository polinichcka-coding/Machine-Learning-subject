```python
def transpose(matrix):
    result=[]
    #* unpack and zip take first element of first row and create tuple
    for row in zip(*matrix):
        result.append(list(row))
    return result

def multiply(A, B):
    result=[]
    for i in range(len(A)):
        row=[]
        for j in range(len(B[0])):
            value=0
            for k in range(len(B)):
                value+=A[i][k]*B[k][j]
            row.append(value)
        result.append(row)
    return result


def inverse(matrix):
    n = len(matrix)
    
    augmented = []
    for i in range(n):
        row = []
        for j in range(n):
            row.append(matrix[i][j])
            
        for j in range(n):
            if i == j:
                row.append(1.0)
            else:
                row.append(0.0)
        augmented.append(row)
        
    for i in range(n):
        max_row = i
        for k in range(i + 1, n):
            if abs(augmented[k][i]) > abs(augmented[max_row][i]):
                max_row = k
        temp_row = augmented[i]
        augmented[i] = augmented[max_row]
        augmented[max_row] = temp_row
        
        pivot = augmented[i][i]
        if abs(pivot) < 1e-12:
            raise ValueError("Cannot be inverted.")
            
        for j in range(2 * n):
            augmented[i][j] = augmented[i][j] / pivot
            
        for k in range(n):
            if k != i:
                factor = augmented[k][i]
                for j in range(2 * n):
                    augmented[k][j] = augmented[k][j] - (factor * augmented[i][j])
                    
    inv_matrix = []
    for i in range(n):
        new_row = []
        for j in range(n, 2 * n):
            new_row.append(augmented[i][j])
        inv_matrix.append(new_row)
        
    return inv_matrix

def ols(X, y): 
    X_T = transpose(X)
    X_T_X = multiply(X_T, X)
    X_T_X_inv = inverse(X_T_X)
    y_matrix = [[value] for value in y]
    X_T_y = multiply(X_T, y_matrix)
    theta = multiply(X_T_X_inv, X_T_y)
    return [row[0] for row in theta]
```
